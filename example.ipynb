```python
import arxiv
```

```python

pdf_loader = arxiv.Search(
    query='cat:cs.AI OR cat:cs.LG  OR cat.cs.CV',
    sort_by=arxiv.SortCriterion.SubmittedDate,
    sort_order=arxiv.SortOrder.Descending,
    max_results=5
)

client = arxiv.Client()
reserch_paper_url = []
for paper in client.results(pdf_loader):

   reserch_paper_url.append(paper.pdf_url)
```

```python
"""
Category	Research area

cs.AI	Artificial Intelligence
cs.LG	Machine Learning
cs.CL	NLP and Computational Linguistics
cs.CV	Computer Vision
cs.NE	Neural and Evolutionary Computing
cs.RO	Robotics
cs.IR	Information Retrieval
cs.SE	Software Engineering
cs.CY	Computers and Society
cs.HC	Human-Computer Interaction
"""
```

```python
import urllib.request

for i in range(len(reserch_paper_url)):
    file_path = f'data/research_paper_{i}.pdf'
    url = reserch_paper_url[i]
    urllib.request.urlretrieve(url, file_path)
```

```python
from langchain_community.document_loaders import PyPDFLoader,PyMuPDFLoader,DirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = DirectoryLoader(
    path='data/pdf',
    glob='**/*.pdf',
    loader_cls=PyMuPDFLoader,
    use_multithreading=True,
    show_progress=True
)

doc = loader.load()

## convert into chunks 
chunk = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=100
)

chunks = chunk.split_documents(doc) 
```
